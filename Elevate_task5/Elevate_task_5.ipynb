```python
df.hist(figsize=(12, 8), bins=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(data=df.select_dtypes(include='number'))
plt.xticks(rotation=90)
plt.show()
```

```python
sns.countplot(x='Sex', data=df)
plt.title('Gender_Distribution')
plt.show()
```

```python
sns.countplot(x='Pclass', data=df)
plt.title('Passenger Class Distribution')
plt.show()
```

```python
sns.countplot(x='Embarked', data=df)
plt.title('Port of Embarked')
plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival by Gender')
plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title('Survival by passanger class')
plt.show()
```

```python
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=df)
plt.title('Age vs fare coloured by survivals')
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(10, 6))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

### Observation:
- Most survivors were female.
- Higher survival rates among 1st class passengers.
- There are missing values in Age and Cabin.

## Summary of Insights:
- Women had a much higher survival rate than men.
- Passengers in 1st class had higher survival chances.
- Age and Fare show interesting patterns with survival.
- Columns like 'Cabin' have many missing values.
